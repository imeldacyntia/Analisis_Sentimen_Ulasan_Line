```python
# Import library yang diperlukan
from google_play_scraper import reviews, Sort
import pandas as pd

# Scraping 30000 review dari aplikasi LINE di Google Play Store
scrapreview = reviews(
    'jp.naver.line.android',
    lang='id',
    country='id',
    sort=Sort.MOST_RELEVANT,
    count=30000,
)

# Mengubah hasil scraping menjadi DataFrame (tabel) menggunakan pandas
data = pd.DataFrame(scrapreview[0])

# Menyimpan hasil scraping ke dalam file CSV
data.to_csv('ulasan_LINE.csv', index=False)

# Menampilkan pesan bahwa proses scraping telah selesai
print("Scraping selesai! File ulasan_LINE.csv telah dibuat.")
```
